```python
# ============================================================
# VGCHARTZ DATASET
# NLP + K-MEANS CLUSTERING
# Complete Jupyter Notebook
# ============================================================


# ============================================================
# CELL 1 - IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")


# ============================================================
# CELL 2 - LOAD DATASET
# ============================================================

FILE_PATH = "vgchartz.csv"

df = pd.read_csv(FILE_PATH)

print("Dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ============================================================
# CELL 3 - DISPLAY FIRST ROWS
# ============================================================

display(df.head(10))


# ============================================================
# CELL 4 - DISPLAY LAST ROWS
# ============================================================

display(df.tail())


# ============================================================
# CELL 5 - DATASET INFORMATION
# ============================================================

print("Dataset Information")
print("=" * 60)

df.info()


# ============================================================
# CELL 6 - COLUMN NAMES
# ============================================================

print("Original column names:")
print(df.columns.tolist())


# ============================================================
# CELL 7 - CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)
)

print("Cleaned column names:")
print(df.columns.tolist())


# ============================================================
# CELL 8 - CHECK DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicates removed.")

print("New dataset shape:", df.shape)


# ============================================================
# CELL 9 - MISSING VALUES
# ============================================================

missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(
    ascending=False
)

print("Missing values:")
display(missing)


# ============================================================
# CELL 10 - MISSING VALUE PERCENTAGES
# ============================================================

missing_percentage = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

missing_table = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": missing_percentage
})

display(
    missing_table[
        missing_table["Missing_Count"] > 0
    ]
)


# ============================================================
# CELL 11 - BASIC STATISTICS
# ============================================================

print("Numerical statistics:")

display(df.describe())


# ============================================================
# CELL 12 - DATA TYPES
# ============================================================

print("Data types:")

display(
    pd.DataFrame({
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str),
        "Unique_Values": [
            df[col].nunique()
            for col in df.columns
        ]
    })
)


# ============================================================
# CELL 13 - IDENTIFY NUMERICAL COLUMNS
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical columns:")
print(numeric_columns)


# ============================================================
# CELL 14 - IDENTIFY CATEGORICAL COLUMNS
# ============================================================

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)


# ============================================================
# CELL 15 - IDENTIFY TEXT COLUMNS FOR NLP
# ============================================================

possible_text_columns = [
    "name",
    "genre",
    "publisher",
    "developer",
    "platform",
    "rating",
    "esrb"
]

text_columns = [
    column
    for column in possible_text_columns
    if column in df.columns
]

print("Text columns selected for NLP:")
print(text_columns)


# ============================================================
# CELL 16 - CREATE TEXT DATA FOR NLP
# ============================================================

if len(text_columns) == 0:
    raise ValueError(
        "No expected text columns were found. "
        "Check df.columns.tolist() and modify text_columns."
    )

df["combined_text"] = (
    df[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
)

display(
    df[
        text_columns + ["combined_text"]
    ].head(10)
)


# ============================================================
# CELL 17 - TEXT CLEANING FUNCTION
# ============================================================

def clean_text(text):

    # Convert to lowercase
    text = str(text).lower()

    # Remove punctuation
    text = re.sub(
        r"[^a-zA-Z0-9\s]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


# ============================================================
# CELL 18 - APPLY TEXT CLEANING
# ============================================================

df["clean_text"] = (
    df["combined_text"]
    .apply(clean_text)
)

display(
    df[
        ["combined_text", "clean_text"]
    ].head(10)
)


# ============================================================
# CELL 19 - CHECK EMPTY TEXT RECORDS
# ============================================================

empty_text = (
    df["clean_text"]
    .str.len() == 0
).sum()

print(
    "Rows with empty text:",
    empty_text
)

if empty_text > 0:

    df = df[
        df["clean_text"].str.len() > 0
    ].copy()

    df.reset_index(
        drop=True,
        inplace=True
    )

print("Dataset shape after text cleaning:", df.shape)


# ============================================================
# CELL 20 - MOST COMMON WORDS
# ============================================================

from collections import Counter

all_words = " ".join(
    df["clean_text"]
).split()

word_counts = Counter(all_words)

common_words = pd.DataFrame(
    word_counts.most_common(30),
    columns=["Word", "Frequency"]
)

display(common_words)


# ============================================================
# CELL 21 - PLOT MOST COMMON WORDS
# ============================================================

plt.figure(figsize=(12, 6))

sns.barplot(
    data=common_words,
    x="Frequency",
    y="Word"
)

plt.title(
    "Top 30 Most Common Words in VGChartz Dataset"
)

plt.xlabel("Frequency")
plt.ylabel("Word")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 22 - TF-IDF VECTORIZATION
# ============================================================

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    min_df=2,
    max_df=0.95,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X_tfidf = tfidf.fit_transform(
    df["clean_text"]
)

print("TF-IDF completed.")
print(
    "Number of documents:",
    X_tfidf.shape[0]
)

print(
    "Number of TF-IDF features:",
    X_tfidf.shape[1]
)


# ============================================================
# CELL 23 - VIEW TF-IDF FEATURE NAMES
# ============================================================

feature_names = (
    tfidf.get_feature_names_out()
)

print(
    "First 50 TF-IDF features:"
)

print(
    feature_names[:50]
)


# ============================================================
# CELL 24 - TF-IDF MATRIX SAMPLE
# ============================================================

tfidf_sample = X_tfidf[
    :10
].toarray()

tfidf_df = pd.DataFrame(
    tfidf_sample,
    columns=feature_names
)

display(
    tfidf_df.iloc[:, :20]
)


# ============================================================
# CELL 25 - FIND OPTIMAL K
# ELBOW METHOD
# ============================================================

k_range = range(2, 11)

inertia_values = []

print("Calculating inertia...")

for k in k_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_tfidf)

    inertia_values.append(
        model.inertia_
    )


# ============================================================
# CELL 26 - PLOT ELBOW CURVE
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_range),
    inertia_values,
    marker="o"
)

plt.title(
    "Elbow Method for Choosing K"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.xticks(
    list(k_range)
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 27 - SILHOUETTE SCORE
# ============================================================

silhouette_values = []

print("Calculating silhouette scores...")

for k in k_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(
        X_tfidf
    )

    score = silhouette_score(
        X_tfidf,
        labels
    )

    silhouette_values.append(score)

    print(
        f"K = {k}, "
        f"Silhouette Score = {score:.4f}"
    )


# ============================================================
# CELL 28 - PLOT SILHOUETTE SCORES
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_range),
    silhouette_values,
    marker="o"
)

plt.title(
    "Silhouette Score for Different K Values"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.xticks(
    list(k_range)
)

plt.grid(True)

plt.show()


# ============================================================
# CELL 29 - SELECT BEST K
# ============================================================

best_k = list(k_range)[
    np.argmax(silhouette_values)
]

best_score = max(
    silhouette_values
)

print(
    "Best K according to silhouette score:",
    best_k
)

print(
    "Best silhouette score:",
    round(best_score, 4)
)


# ============================================================
# CELL 30 - K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10,
    max_iter=300
)

cluster_labels = kmeans.fit_predict(
    X_tfidf
)

df["cluster"] = cluster_labels

print(
    "K-Means clustering completed."
)


# ============================================================
# CELL 31 - CLUSTER COUNTS
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("Games in each cluster:")

display(
    cluster_counts.to_frame(
        name="Number_of_Games"
    )
)


# ============================================================
# CELL 32 - CLUSTER SIZE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="cluster"
)

plt.title(
    "Number of Games in Each Cluster"
)

plt.xlabel(
    "Cluster"
)

plt.ylabel(
    "Number of Games"
)

plt.show()


# ============================================================
# CELL 33 - MOST IMPORTANT TERMS IN EACH CLUSTER
# ============================================================

cluster_centers = (
    kmeans.cluster_centers_
)

for cluster_number in range(best_k):

    print("\n")
    print("=" * 70)
    print(
        f"CLUSTER {cluster_number}"
    )
    print("=" * 70)

    top_indices = (
        cluster_centers[
            cluster_number
        ]
        .argsort()[-20:][::-1]
    )

    top_terms = [
        feature_names[index]
        for index in top_indices
    ]

    print(
        ", ".join(top_terms)
    )


# ============================================================
# CELL 34 - CREATE CLUSTER TERM TABLE
# ============================================================

cluster_terms = []

for cluster_number in range(best_k):

    top_indices = (
        cluster_centers[
            cluster_number
        ]
        .argsort()[-15:][::-1]
    )

    terms = [
        feature_names[index]
        for index in top_indices
    ]

    cluster_terms.append({
        "Cluster": cluster_number,
        "Important_Terms": ", ".join(terms)
    })

cluster_terms_df = pd.DataFrame(
    cluster_terms
)

display(
    cluster_terms_df
)


# ============================================================
# CELL 35 - DISPLAY EXAMPLES FROM EACH CLUSTER
# ============================================================

display_columns = [
    column
    for column in [
        "name",
        "platform",
        "genre",
        "publisher",
        "developer",
        "cluster"
    ]
    if column in df.columns
]

for cluster_number in range(best_k):

    print("\n")
    print("=" * 70)
    print(
        f"EXAMPLES FROM CLUSTER {cluster_number}"
    )
    print("=" * 70)

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    display(
        cluster_data[
            display_columns
        ].head(20)
    )


# ============================================================
# CELL 36 - PCA REDUCTION
# ============================================================

print(
    "Reducing TF-IDF dimensions using PCA..."
)

# PCA requires dense data.
# For a very large dataset, use TruncatedSVD instead.

X_dense = X_tfidf.toarray()

pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(
    X_dense
)

print(
    "Explained variance:",
    pca.explained_variance_ratio_
)

print(
    "Total explained variance:",
    round(
        pca.explained_variance_ratio_.sum(),
        4
    )
)


# ============================================================
# CELL 37 - ADD PCA COLUMNS
# ============================================================

df["PCA1"] = X_pca[:, 0]
df["PCA2"] = X_pca[:, 1]


# ============================================================
# CELL 38 - PCA CLUSTER VISUALISATION
# ============================================================

plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=df,
    x="PCA1",
    y="PCA2",
    hue="cluster",
    palette="tab10",
    alpha=0.7,
    s=60
)

plt.title(
    "K-Means Clusters Using TF-IDF + PCA"
)

plt.xlabel(
    "PCA Component 1"
)

plt.ylabel(
    "PCA Component 2"
)

plt.legend(
    title="Cluster",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 39 - GENRE ANALYSIS
# ============================================================

if "genre" in df.columns:

    print(
        "Genre distribution:"
    )

    genre_counts = (
        df["genre"]
        .value_counts()
    )

    display(
        genre_counts.to_frame(
            name="Number_of_Games"
        ).head(20)
    )


# ============================================================
# CELL 40 - GENRE DISTRIBUTION GRAPH
# ============================================================

if "genre" in df.columns:

    top_genres = (
        df["genre"]
        .value_counts()
        .head(15)
        .index
    )

    genre_plot_data = df[
        df["genre"].isin(top_genres)
    ]

    plt.figure(figsize=(12, 7))

    sns.countplot(
        data=genre_plot_data,
        y="genre",
        order=top_genres
    )

    plt.title(
        "Top 15 Game Genres"
    )

    plt.xlabel(
        "Number of Games"
    )

    plt.ylabel(
        "Genre"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# CELL 41 - GENRE BY CLUSTER
# ============================================================

if "genre" in df.columns:

    genre_cluster = pd.crosstab(
        df["cluster"],
        df["genre"]
    )

    display(
        genre_cluster
    )


# ============================================================
# CELL 42 - GENRE/CLUSTER HEATMAP
# ============================================================

if "genre" in df.columns:

    plt.figure(figsize=(16, 8))

    sns.heatmap(
        genre_cluster,
        annot=True,
        fmt="d",
        cmap="Blues"
    )

    plt.title(
        "Genre Distribution by K-Means Cluster"
    )

    plt.xlabel(
        "Genre"
    )

    plt.ylabel(
        "Cluster"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# CELL 43 - PLATFORM ANALYSIS
# ============================================================

if "platform" in df.columns:

    platform_counts = (
        df["platform"]
        .value_counts()
        .head(20)
    )

    display(
        platform_counts.to_frame(
            name="Number_of_Games"
        )
    )


# ============================================================
# CELL 44 - PLATFORM BY CLUSTER
# ============================================================

if "platform" in df.columns:

    platform_cluster = pd.crosstab(
        df["cluster"],
        df["platform"]
    )

    display(
        platform_cluster
    )


# ============================================================
# CELL 45 - PUBLISHER ANALYSIS
# ============================================================

if "publisher" in df.columns:

    publisher_counts = (
        df["publisher"]
        .value_counts()
        .head(20)
    )

    display(
        publisher_counts.to_frame(
            name="Number_of_Games"
        )
    )


# ============================================================
# CELL 46 - SALES COLUMNS
# ============================================================

sales_columns = [
    column
    for column in df.columns
    if "sales" in column.lower()
]

print(
    "Potential sales columns:"
)

print(
    sales_columns
)


# ============================================================
# CELL 47 - CONVERT SALES COLUMNS TO NUMERIC
# ============================================================

for column in sales_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print(
    "Sales columns converted where possible."
)


# ============================================================
# CELL 48 - SALES BY CLUSTER
# ============================================================

numeric_sales_columns = [
    column
    for column in sales_columns
    if pd.api.types.is_numeric_dtype(
        df[column]
    )
]

if len(numeric_sales_columns) > 0:

    sales_by_cluster = (
        df.groupby("cluster")[
            numeric_sales_columns
        ]
        .mean()
        .round(2)
    )

    print(
        "Average sales by cluster:"
    )

    display(
        sales_by_cluster
    )

else:

    print(
        "No numeric sales columns were found."
    )


# ============================================================
# CELL 49 - GLOBAL SALES BY CLUSTER
# ============================================================

global_sales_candidates = [
    column
    for column in [
        "global_sales",
        "global_sales_m",
        "global_sales_millions"
    ]
    if column in df.columns
]

if len(global_sales_candidates) > 0:

    sales_column = (
        global_sales_candidates[0]
    )

    cluster_sales = (
        df.groupby("cluster")[
            sales_column
        ]
        .agg([
            "count",
            "mean",
            "median",
            "sum"
        ])
        .round(2)
    )

    display(
        cluster_sales
    )


# ============================================================
# CELL 50 - TOP SELLING GAMES
# ============================================================

if len(global_sales_candidates) > 0:

    sales_column = (
        global_sales_candidates[0]
    )

    top_games = (
        df.sort_values(
            by=sales_column,
            ascending=False
        )
        .head(20)
    )

    columns_for_display = [
        column
        for column in [
            "name",
            "platform",
            "genre",
            "publisher",
            sales_column,
            "cluster"
        ]
        if column in df.columns
    ]

    display(
        top_games[
            columns_for_display
        ]
    )


# ============================================================
# CELL 51 - SALES DISTRIBUTION
# ============================================================

if len(global_sales_candidates) > 0:

    sales_column = (
        global_sales_candidates[0]
    )

    plt.figure(figsize=(10, 6))

    sns.histplot(
        df[sales_column].dropna(),
        bins=40,
        kde=True
    )

    plt.title(
        "Distribution of Global Game Sales"
    )

    plt.xlabel(
        sales_column
    )

    plt.ylabel(
        "Number of Games"
    )

    plt.show()


# ============================================================
# CELL 52 - CLUSTER VS SALES BOX PLOT
# ============================================================

if len(global_sales_candidates) > 0:

    sales_column = (
        global_sales_candidates[0]
    )

    plt.figure(figsize=(12, 7))

    sns.boxplot(
        data=df,
        x="cluster",
        y=sales_column
    )

    plt.title(
        "Sales Distribution by K-Means Cluster"
    )

    plt.xlabel(
        "Cluster"
    )

    plt.ylabel(
        sales_column
    )

    plt.show()


# ============================================================
# CELL 53 - CLUSTER SUMMARY
# ============================================================

summary_data = []

for cluster_number in range(best_k):

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    summary = {
        "Cluster": cluster_number,
        "Number_of_Games": len(cluster_data)
    }

    if "genre" in df.columns:

        most_common_genre = (
            cluster_data["genre"]
            .value_counts()
            .index[0]
            if len(cluster_data) > 0
            else "N/A"
        )

        summary[
            "Most_Common_Genre"
        ] = most_common_genre

    if "platform" in df.columns:

        most_common_platform = (
            cluster_data["platform"]
            .value_counts()
            .index[0]
            if len(cluster_data) > 0
            else "N/A"
        )

        summary[
            "Most_Common_Platform"
        ] = most_common_platform

    if "publisher" in df.columns:

        most_common_publisher = (
            cluster_data["publisher"]
            .value_counts()
            .index[0]
            if len(cluster_data) > 0
            else "N/A"
        )

        summary[
            "Most_Common_Publisher"
        ] = most_common_publisher

    if len(global_sales_candidates) > 0:

        sales_column = (
            global_sales_candidates[0]
        )

        summary[
            "Average_Sales"
        ] = cluster_data[
            sales_column
        ].mean()

    summary_data.append(
        summary
    )

cluster_summary = pd.DataFrame(
    summary_data
)

display(
    cluster_summary
)


# ============================================================
# CELL 54 - FIND CLOSEST GAMES TO EACH CLUSTER CENTRE
# ============================================================

# Distance between each game and its cluster centre

distances = kmeans.transform(
    X_tfidf
)

df["distance_to_cluster"] = [
    distances[i, cluster]
    for i, cluster
    in enumerate(
        df["cluster"]
    )
]


# ============================================================
# CELL 55 - MOST REPRESENTATIVE GAMES
# ============================================================

for cluster_number in range(best_k):

    print("\n")
    print("=" * 70)
    print(
        f"MOST REPRESENTATIVE GAMES - CLUSTER {cluster_number}"
    )
    print("=" * 70)

    representative_games = (
        df[
            df["cluster"] == cluster_number
        ]
        .sort_values(
            "distance_to_cluster"
        )
        .head(10)
    )

    columns = [
        column
        for column in [
            "name",
            "platform",
            "genre",
            "publisher",
            "developer",
            "cluster",
            "distance_to_cluster"
        ]
        if column in df.columns
    ]

    display(
        representative_games[
            columns
        ]
    )


# ============================================================
# CELL 56 - PREDICT CLUSTER FOR A NEW GAME
# ============================================================

new_game = pd.DataFrame({
    "name": [
        "Example Adventure Game"
    ],
    "genre": [
        "Action Adventure"
    ],
    "publisher": [
        "Example Publisher"
    ],
    "developer": [
        "Example Developer"
    ],
    "platform": [
        "PS4"
    ]
})


# Add missing columns if required
for column in text_columns:

    if column not in new_game.columns:
        new_game[column] = ""


# Create combined text
new_game["combined_text"] = (
    new_game[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
)


# Clean text
new_game["clean_text"] = (
    new_game["combined_text"]
    .apply(clean_text)
)


# Transform using existing TF-IDF model
new_game_tfidf = (
    tfidf.transform(
        new_game["clean_text"]
    )
)


# Predict cluster
predicted_cluster = (
    kmeans.predict(
        new_game_tfidf
    )
)


print(
    "Predicted cluster for new game:",
    predicted_cluster[0]
)


# ============================================================
# CELL 57 - SAVE COMPLETE DATASET
# ============================================================

OUTPUT_FILE = (
    "vgchartz_kmeans_results.csv"
)

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    f"Saved clustered dataset to: {OUTPUT_FILE}"
)


# ============================================================
# CELL 58 - SAVE CLUSTER SUMMARY
# ============================================================

cluster_summary.to_csv(
    "vgchartz_cluster_summary.csv",
    index=False
)

print(
    "Saved cluster summary."
)


# ============================================================
# CELL 59 - SAVE CLUSTER TERMS
# ============================================================

cluster_terms_df.to_csv(
    "vgchartz_cluster_terms.csv",
    index=False
)

print(
    "Saved cluster terms."
)


# ============================================================
# CELL 60 - FINAL RESULTS
# ============================================================

print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(
    "Original dataset rows:",
    df.shape[0]
)

print(
    "Number of TF-IDF features:",
    X_tfidf.shape[1]
)

print(
    "Number of clusters:",
    best_k
)

print(
    "Silhouette score:",
    round(best_score, 4)
)

print(
    "\nCluster sizes:"
)

display(
    cluster_counts.to_frame(
        name="Games"
    )
)

print(
    "\nCluster summary:"
)

display(
    cluster_summary
)

print(
    "\nFiles created:"
)

print(
    "1. vgchartz_kmeans_results.csv"
)

print(
    "2. vgchartz_cluster_summary.csv"
)

print(
    "3. vgchartz_cluster_terms.csv"
)

print(
    "\nAnalysis completed successfully."
)
```
