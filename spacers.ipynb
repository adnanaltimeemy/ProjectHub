```python
# ============================================================
# SPACE MISSION DATASET
# NLP + K-MEANS CLUSTERING
# Full Jupyter Notebook Code
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
import string
import warnings

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

warnings.filterwarnings("ignore")

print("All libraries imported successfully.")


# ============================================================
# 2. LOAD DATASET
# ============================================================

file_path = "space-mission-data.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print()
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 3. BASIC DATASET INFORMATION
# ============================================================

print("Dataset information:")
print("=" * 60)

df.info()

print("\nMissing values:")
display(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())


# ============================================================
# 4. REMOVE DUPLICATE ROWS
# ============================================================

df = df.drop_duplicates()

df = df.reset_index(drop=True)

print("Dataset shape after removing duplicates:")
print(df.shape)


# ============================================================
# 5. FIND TEXT COLUMNS
# ============================================================

print("\nText/Object columns:")

text_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print(text_columns)


# ============================================================
# 6. SELECT TEXT COLUMN FOR NLP
# ============================================================

# The program tries to automatically find a suitable
# text column.

possible_text_columns = [
    "description",
    "mission_description",
    "mission description",
    "summary",
    "details",
    "objective",
    "objectives",
    "mission",
    "mission_name",
    "mission name",
    "name",
    "title",
    "purpose"
]

TEXT_COLUMN = None

# First search for likely column names
for column in possible_text_columns:

    if column in df.columns:
        TEXT_COLUMN = column
        break


# If no suitable column was found,
# automatically select the first text column.
if TEXT_COLUMN is None:

    if len(text_columns) > 0:

        TEXT_COLUMN = text_columns[0]

    else:

        raise ValueError(
            "No text column found in the dataset. "
            "Please create a text column for NLP."
        )


print("\nSelected NLP text column:")
print(TEXT_COLUMN)


# ============================================================
# 7. CREATE A COPY OF THE TEXT DATA
# ============================================================

df["original_text"] = df[TEXT_COLUMN].fillna("").astype(str)

print("\nExample text:")
display(df[[TEXT_COLUMN, "original_text"]].head())


# ============================================================
# 8. TEXT CLEANING FUNCTION
# ============================================================

def clean_text(text):

    # Convert to string
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )

    # Remove email addresses
    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # Remove HTML tags
    text = re.sub(
        r"<.*?>",
        " ",
        text
    )

    # Remove numbers
    text = re.sub(
        r"\d+",
        " ",
        text
    )

    # Remove punctuation
    text = text.translate(
        str.maketrans(
            "",
            "",
            string.punctuation
        )
    )

    # Remove extra whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Remove leading/trailing spaces
    text = text.strip()

    return text


# ============================================================
# 9. CLEAN TEXT
# ============================================================

df["clean_text"] = df["original_text"].apply(
    clean_text
)

print("Text cleaning completed.")

display(
    df[
        [
            "original_text",
            "clean_text"
        ]
    ].head(10)
)


# ============================================================
# 10. REMOVE EMPTY TEXT ROWS
# ============================================================

before = len(df)

df = df[
    df["clean_text"].str.len() > 0
].copy()

df = df.reset_index(drop=True)

after = len(df)

print("Rows before removing empty text:", before)
print("Rows after removing empty text:", after)
print("Rows removed:", before - after)


# ============================================================
# 11. NLP - TF-IDF VECTORIZATION
# ============================================================

vectorizer = TfidfVectorizer(

    # Remove common English words
    stop_words="english",

    # Use individual words and two-word phrases
    ngram_range=(1, 2),

    # Ignore extremely rare words
    min_df=1,

    # Ignore terms appearing in more than 95% of documents
    max_df=0.95,

    # Limit vocabulary size
    max_features=5000,

    # Normalise TF-IDF vectors
    norm="l2"
)


X = vectorizer.fit_transform(
    df["clean_text"]
)


print("TF-IDF vectorisation completed.")

print("\nTF-IDF matrix shape:")
print(X.shape)

print("\nNumber of TF-IDF features:")
print(len(vectorizer.get_feature_names_out()))


# ============================================================
# 12. DISPLAY TF-IDF FEATURES
# ============================================================

feature_names = vectorizer.get_feature_names_out()

print("\nFirst 50 TF-IDF features:")

for feature in feature_names[:50]:
    print(feature)


# ============================================================
# 13. FIND MOST IMPORTANT WORDS IN THE DATASET
# ============================================================

mean_tfidf = np.asarray(
    X.mean(axis=0)
).flatten()

top_indices = mean_tfidf.argsort()[
    ::-1
][:30]

top_words = pd.DataFrame({

    "Word/Phrase":
        feature_names[top_indices],

    "Average TF-IDF":
        mean_tfidf[top_indices]

})

print("\nTop 30 words/phrases:")
display(top_words)


# ============================================================
# 14. PLOT TOP TF-IDF WORDS
# ============================================================

plt.figure(figsize=(12, 7))

plt.barh(
    top_words["Word/Phrase"][::-1],
    top_words["Average TF-IDF"][::-1]
)

plt.xlabel("Average TF-IDF Score")
plt.ylabel("Word / Phrase")
plt.title(
    "Top 30 Important Words/Phrases in Space Mission Dataset"
)

plt.tight_layout()
plt.show()


# ============================================================
# 15. CHECK DATASET SIZE
# ============================================================

number_of_documents = X.shape[0]

print("Number of documents:", number_of_documents)

if number_of_documents < 3:

    raise ValueError(
        "There are not enough text records for K-Means clustering."
    )


# ============================================================
# 16. FIND BEST NUMBER OF CLUSTERS
# ============================================================

# We test K values from 2 to 10.
# If the dataset is smaller, the maximum K is reduced.

max_k = min(
    10,
    number_of_documents - 1
)

k_values = range(
    2,
    max_k + 1
)

silhouette_scores = []

print("\nTesting different K values:")
print("=" * 60)

for k in k_values:

    model = KMeans(

        n_clusters=k,

        random_state=42,

        n_init=10,

        max_iter=300
    )

    labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k:2d}   "
        f"Silhouette Score = {score:.4f}"
    )


# ============================================================
# 17. CREATE SILHOUETTE SCORE TABLE
# ============================================================

silhouette_df = pd.DataFrame({

    "Number of Clusters (K)":
        list(k_values),

    "Silhouette Score":
        silhouette_scores

})

print("\nSilhouette scores:")
display(silhouette_df)


# ============================================================
# 18. PLOT SILHOUETTE SCORES
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o",
    linewidth=2
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.title(
    "Silhouette Score for Different K Values"
)

plt.xticks(
    list(k_values)
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 19. SELECT BEST K
# ============================================================

best_index = np.argmax(
    silhouette_scores
)

best_k = list(k_values)[best_index]

best_score = silhouette_scores[best_index]

print(
    f"Best K = {best_k}"
)

print(
    f"Best Silhouette Score = {best_score:.4f}"
)


# ============================================================
# 20. APPLY FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(

    n_clusters=best_k,

    random_state=42,

    n_init=10,

    max_iter=300
)

df["cluster"] = kmeans.fit_predict(X)

print("K-Means clustering completed.")


# ============================================================
# 21. DISPLAY NUMBER OF MISSIONS PER CLUSTER
# ============================================================

cluster_counts = (
    df["cluster"]
    .value_counts()
    .sort_index()
)

print("\nNumber of missions in each cluster:")

display(
    cluster_counts.to_frame(
        "Number of Missions"
    )
)


# ============================================================
# 22. PLOT NUMBER OF MISSIONS PER CLUSTER
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.xlabel("Cluster")
plt.ylabel("Number of Missions")

plt.title(
    "Number of Space Missions in Each Cluster"
)

plt.tight_layout()

plt.show()


# ============================================================
# 23. FIND IMPORTANT TERMS FOR EACH CLUSTER
# ============================================================

order_centroids = (
    kmeans.cluster_centers_
    .argsort()[:, ::-1]
)


cluster_keywords = {}

print("\nImportant terms for each cluster:")
print("=" * 70)

for cluster_number in range(best_k):

    top_indices_cluster = (
        order_centroids[
            cluster_number,
            :15
        ]
    )

    terms = [
        feature_names[i]
        for i in top_indices_cluster
    ]

    cluster_keywords[
        cluster_number
    ] = terms

    print(
        f"\nCluster {cluster_number}:"
    )

    print(
        ", ".join(terms)
    )


# ============================================================
# 24. CREATE CLUSTER KEYWORDS TABLE
# ============================================================

cluster_keywords_df = pd.DataFrame({

    "Cluster":
        list(cluster_keywords.keys()),

    "Top Keywords":
        [
            ", ".join(
                cluster_keywords[c]
            )
            for c in cluster_keywords
        ]

})

display(
    cluster_keywords_df
)


# ============================================================
# 25. DISPLAY EXAMPLES FROM EACH CLUSTER
# ============================================================

print("\nExample missions from each cluster:")
print("=" * 70)

for cluster_number in range(best_k):

    print(
        f"\nCLUSTER {cluster_number}"
    )

    print("-" * 70)

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    display(
        cluster_data[
            [
                TEXT_COLUMN,
                "cluster"
            ]
        ].head(10)
    )


# ============================================================
# 26. PCA FOR VISUALISATION
# ============================================================

# Convert the TF-IDF matrix to a dense matrix.
# PCA will reduce the data to 2 dimensions.

X_dense = X.toarray()

print(
    "Dense matrix shape:",
    X_dense.shape
)


# ============================================================
# 27. APPLY PCA
# ============================================================

pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(
    X_dense
)

print(
    "PCA completed."
)

print(
    "Explained variance ratio:",
    pca.explained_variance_ratio_
)

print(
    "Total explained variance:",
    pca.explained_variance_ratio_.sum()
)


# ============================================================
# 28. ADD PCA VALUES TO DATAFRAME
# ============================================================

df["PCA_1"] = X_pca[:, 0]

df["PCA_2"] = X_pca[:, 1]


# ============================================================
# 29. VISUALISE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(12, 8))

scatter = plt.scatter(

    df["PCA_1"],

    df["PCA_2"],

    c=df["cluster"],

    s=60,

    alpha=0.7
)

plt.xlabel(
    "Principal Component 1"
)

plt.ylabel(
    "Principal Component 2"
)

plt.title(
    "K-Means Clustering of Space Missions"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 30. CALCULATE FINAL SILHOUETTE SCORE
# ============================================================

final_silhouette = silhouette_score(
    X,
    df["cluster"]
)

print(
    "Final Silhouette Score:",
    round(final_silhouette, 4)
)


# ============================================================
# 31. CREATE CLUSTER SUMMARY
# ============================================================

cluster_summary = []

for cluster_number in range(best_k):

    cluster_data = df[
        df["cluster"] == cluster_number
    ]

    keywords = cluster_keywords[
        cluster_number
    ]

    cluster_summary.append({

        "Cluster":
            cluster_number,

        "Number of Missions":
            len(cluster_data),

        "Percentage of Dataset":
            round(
                len(cluster_data)
                / len(df)
                * 100,
                2
            ),

        "Main Keywords":
            ", ".join(keywords[:10])

    })


cluster_summary_df = pd.DataFrame(
    cluster_summary
)


# ============================================================
# 32. DISPLAY CLUSTER SUMMARY
# ============================================================

print("\nCLUSTER SUMMARY")
print("=" * 70)

display(
    cluster_summary_df
)


# ============================================================
# 33. CREATE HUMAN-READABLE CLUSTER LABELS
# ============================================================

# Automatically create labels using the first
# few important terms for each cluster.

cluster_labels = {}

for cluster_number in range(best_k):

    keywords = cluster_keywords[
        cluster_number
    ]

    label = " / ".join(
        keywords[:3]
    )

    cluster_labels[
        cluster_number
    ] = label


df["cluster_label"] = df[
    "cluster"
].map(cluster_labels)


# ============================================================
# 34. DISPLAY CLUSTER LABELS
# ============================================================

print("\nAutomatically generated cluster labels:")

for cluster_number, label in cluster_labels.items():

    print(
        f"Cluster {cluster_number}: {label}"
    )


# ============================================================
# 35. SHOW FINAL DATASET
# ============================================================

print("\nFinal dataset:")
display(
    df.head(20)
)


# ============================================================
# 36. SAVE CLUSTERED DATASET
# ============================================================

output_file = (
    "space-mission-data-clustered.csv"
)

df.to_csv(
    output_file,
    index=False
)

print(
    "\nClustered dataset saved successfully:"
)

print(
    output_file
)


# ============================================================
# 37. SAVE CLUSTER SUMMARY
# ============================================================

summary_file = (
    "space-mission-cluster-summary.csv"
)

cluster_summary_df.to_csv(
    summary_file,
    index=False
)

print(
    "Cluster summary saved successfully:"
)

print(
    summary_file
)


# ============================================================
# 38. FINAL RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(
    f"Original number of records: "
    f"{before}"
)

print(
    f"Records used for NLP: "
    f"{len(df)}"
)

print(
    f"Text column used: "
    f"{TEXT_COLUMN}"
)

print(
    f"Number of TF-IDF features: "
    f"{X.shape[1]}"
)

print(
    f"Optimal number of clusters: "
    f"{best_k}"
)

print(
    f"Silhouette Score: "
    f"{final_silhouette:.4f}"
)

print("\nCluster sizes:")

for cluster_number in range(best_k):

    count = len(
        df[
            df["cluster"] == cluster_number
        ]
    )

    print(
        f"Cluster {cluster_number}: "
        f"{count} missions"
    )

print("\nTop keywords by cluster:")

for cluster_number in range(best_k):

    print(
        f"\nCluster {cluster_number}:"
    )

    print(
        ", ".join(
            cluster_keywords[
                cluster_number
            ][:10]
        )
    )

print("\nAnalysis completed successfully!")
```
