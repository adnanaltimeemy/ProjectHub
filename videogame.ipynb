```python
# ============================================================
# VIDEO GAME SALES ANALYSIS
# NLP + TF-IDF + K-MEANS CLUSTERING
# Dataset:
# Video_Games_Sales_as_at_22_Dec_2016.csv
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")


# ============================================================
# 2. LOAD DATASET
# ============================================================

file_path = "Video_Games_Sales_as_at_22_Dec_2016.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)

display(df.head())


# ============================================================
# 3. INITIAL DATA EXPLORATION
# ============================================================

print("\n================ DATASET INFORMATION ================\n")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst 10 rows:")
display(df.head(10))


# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\n================ MISSING VALUES ================\n")

missing_values = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})

display(
    missing_table.sort_values(
        by="Missing Values",
        ascending=False
    )
)


# ============================================================
# 5. CHECK DUPLICATES
# ============================================================

print("\nNumber of duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)


# ============================================================
# 6. DATA CLEANING
# ============================================================

# Remove rows where game Name is missing
df = df.dropna(subset=["Name"]).copy()

# Convert Name to string
df["Name"] = df["Name"].astype(str)

# Convert numerical sales columns to numeric
sales_columns = [
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales",
    "Global_Sales"
]

for column in sales_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

# Fill missing sales values with zero
for column in sales_columns:
    if column in df.columns:
        df[column] = df[column].fillna(0)

print("Data cleaning completed.")
print("Final dataset shape:", df.shape)


# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

print("\n================ DESCRIPTIVE STATISTICS ================\n")

display(df.describe(include="all"))


# ============================================================
# 8. BASIC DATA VISUALISATION
# ============================================================

# Top 20 games by global sales

top_games = (
    df.sort_values(
        by="Global_Sales",
        ascending=False
    )
    .head(20)
)

print("\nTop 20 Games by Global Sales:")
display(
    top_games[
        [
            "Name",
            "Platform",
            "Year_of_Release",
            "Genre",
            "Global_Sales"
        ]
    ]
)


plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_games,
    x="Global_Sales",
    y="Name"
)

plt.title("Top 20 Video Games by Global Sales")
plt.xlabel("Global Sales (Millions)")
plt.ylabel("Game")

plt.tight_layout()
plt.show()


# ============================================================
# 9. SALES BY GENRE
# ============================================================

genre_sales = (
    df.groupby("Genre")["Global_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nGlobal Sales by Genre:")
display(genre_sales)


plt.figure(figsize=(12, 6))

genre_sales.plot(kind="bar")

plt.title("Total Global Sales by Genre")
plt.xlabel("Genre")
plt.ylabel("Global Sales (Millions)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 10. SALES BY PLATFORM
# ============================================================

platform_sales = (
    df.groupby("Platform")["Global_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(20)
)

print("\nTop 20 Platforms by Global Sales:")
display(platform_sales)


plt.figure(figsize=(12, 6))

platform_sales.plot(kind="bar")

plt.title("Top 20 Platforms by Global Sales")
plt.xlabel("Platform")
plt.ylabel("Global Sales (Millions)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 11. NLP TEXT PREPROCESSING
# ============================================================

def clean_text(text):

    # Convert text to lowercase
    text = text.lower()

    # Remove special characters
    text = re.sub(
        r"[^a-zA-Z0-9\s]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Remove leading/trailing spaces
    text = text.strip()

    return text


df["Name_Clean"] = df["Name"].apply(clean_text)

print("\nOriginal and cleaned game names:")

display(
    df[
        [
            "Name",
            "Name_Clean"
        ]
    ].head(20)
)


# ============================================================
# 12. TF-IDF VECTORIZATION
# ============================================================

print("\n================ TF-IDF ================\n")

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    ngram_range=(1, 2),
    min_df=2
)

X_tfidf = tfidf.fit_transform(
    df["Name_Clean"]
)

print("TF-IDF matrix shape:")
print(X_tfidf.shape)

print(
    "Number of TF-IDF features:",
    len(tfidf.get_feature_names_out())
)


# ============================================================
# 13. DISPLAY IMPORTANT TF-IDF WORDS
# ============================================================

feature_names = tfidf.get_feature_names_out()

print("\nFirst 100 TF-IDF features:")
print(feature_names[:100])


# ============================================================
# 14. ELBOW METHOD
# ============================================================

print("\n================ ELBOW METHOD ================\n")

inertia = []

k_values = range(2, 11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_tfidf)

    inertia.append(
        model.inertia_
    )


plt.figure(figsize=(10, 6))

plt.plot(
    k_values,
    inertia,
    marker="o"
)

plt.title("Elbow Method for K-Means")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(k_values)
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 15. SILHOUETTE SCORE
# ============================================================

print("\n================ SILHOUETTE ANALYSIS ================\n")

silhouette_scores = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(
        X_tfidf
    )

    score = silhouette_score(
        X_tfidf,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k}: Silhouette Score = {score:.4f}"
    )


plt.figure(figsize=(10, 6))

plt.plot(
    k_values,
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score for K-Means")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(k_values)
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 16. AUTOMATICALLY SELECT BEST K
# ============================================================

best_k = k_values[
    np.argmax(silhouette_scores)
]

best_score = max(
    silhouette_scores
)

print("\nBest K according to silhouette score:", best_k)
print("Best silhouette score:", round(best_score, 4))


# ============================================================
# 17. TRAIN FINAL K-MEANS MODEL
# ============================================================

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(
    X_tfidf
)

print("\nK-Means clustering completed.")

display(
    df[
        [
            "Name",
            "Genre",
            "Platform",
            "Global_Sales",
            "Cluster"
        ]
    ].head(20)
)


# ============================================================
# 18. NUMBER OF GAMES IN EACH CLUSTER
# ============================================================

cluster_counts = (
    df["Cluster"]
    .value_counts()
    .sort_index()
)

print("\nNumber of games in each cluster:")
display(cluster_counts)


plt.figure(figsize=(10, 6))

cluster_counts.plot(
    kind="bar"
)

plt.title("Number of Games in Each Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Games")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 19. FIND IMPORTANT WORDS FOR EACH CLUSTER
# ============================================================

print("\n================ CLUSTER WORDS ================\n")

terms = tfidf.get_feature_names_out()

order_centroids = (
    kmeans.cluster_centers_
    .argsort()[:, ::-1]
)

for cluster in range(best_k):

    print("\n----------------------------------------")
    print(f"CLUSTER {cluster}")
    print("----------------------------------------")

    top_terms = []

    for index in order_centroids[
        cluster,
        :20
    ]:

        top_terms.append(
            terms[index]
        )

    print(
        ", ".join(top_terms)
    )


# ============================================================
# 20. SHOW TOP GAMES IN EACH CLUSTER
# ============================================================

print("\n================ GAMES BY CLUSTER ================\n")

for cluster in sorted(
    df["Cluster"].unique()
):

    print("\n")
    print("=" * 70)
    print(f"CLUSTER {cluster}")
    print("=" * 70)

    cluster_games = (
        df[
            df["Cluster"] == cluster
        ]
        .sort_values(
            by="Global_Sales",
            ascending=False
        )
        .head(15)
    )

    display(
        cluster_games[
            [
                "Name",
                "Platform",
                "Genre",
                "Year_of_Release",
                "Global_Sales"
            ]
        ]
    )


# ============================================================
# 21. CLUSTER SALES ANALYSIS
# ============================================================

cluster_sales = (
    df.groupby("Cluster")[
        [
            "NA_Sales",
            "EU_Sales",
            "JP_Sales",
            "Other_Sales",
            "Global_Sales"
        ]
    ]
    .agg(["mean", "sum"])
)

print("\n================ CLUSTER SALES ================\n")

display(cluster_sales)


# ============================================================
# 22. AVERAGE GLOBAL SALES BY CLUSTER
# ============================================================

average_sales = (
    df.groupby("Cluster")[
        "Global_Sales"
    ]
    .mean()
    .sort_values(
        ascending=False
    )
)

print("\nAverage Global Sales by Cluster:")
display(average_sales)


plt.figure(figsize=(10, 6))

average_sales.plot(
    kind="bar"
)

plt.title("Average Global Sales by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Average Global Sales (Millions)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 23. TOTAL GLOBAL SALES BY CLUSTER
# ============================================================

total_sales = (
    df.groupby("Cluster")[
        "Global_Sales"
    ]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\nTotal Global Sales by Cluster:")
display(total_sales)


plt.figure(figsize=(10, 6))

total_sales.plot(
    kind="bar"
)

plt.title("Total Global Sales by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Total Global Sales (Millions)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 24. GENRE DISTRIBUTION BY CLUSTER
# ============================================================

genre_cluster = pd.crosstab(
    df["Cluster"],
    df["Genre"]
)

print("\nGenre distribution by cluster:")
display(genre_cluster)


plt.figure(figsize=(14, 7))

sns.heatmap(
    genre_cluster,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Genre Distribution Across K-Means Clusters"
)

plt.xlabel("Genre")
plt.ylabel("Cluster")

plt.tight_layout()
plt.show()


# ============================================================
# 25. PLATFORM DISTRIBUTION BY CLUSTER
# ============================================================

platform_cluster = pd.crosstab(
    df["Cluster"],
    df["Platform"]
)

print("\nPlatform distribution by cluster:")
display(platform_cluster)


# ============================================================
# 26. REDUCE TF-IDF TO 2 DIMENSIONS
# ============================================================

print("\n================ DIMENSION REDUCTION ================\n")

svd = TruncatedSVD(
    n_components=2,
    random_state=42
)

X_2d = svd.fit_transform(
    X_tfidf
)

print(
    "Original TF-IDF dimensions:",
    X_tfidf.shape
)

print(
    "Reduced dimensions:",
    X_2d.shape
)


# ============================================================
# 27. VISUALIZE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=df["Cluster"],
    cmap="viridis",
    alpha=0.6,
    s=30
)

plt.title(
    "K-Means Clustering of Video Game Names"
)

plt.xlabel(
    "SVD Component 1"
)

plt.ylabel(
    "SVD Component 2"
)

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 28. ADD SVD COMPONENTS TO DATAFRAME
# ============================================================

df["SVD_Component_1"] = X_2d[:, 0]
df["SVD_Component_2"] = X_2d[:, 1]


# ============================================================
# 29. CLUSTER SUMMARY
# ============================================================

cluster_summary = (
    df.groupby("Cluster")
    .agg(
        Number_of_Games=("Name", "count"),
        Average_Global_Sales=(
            "Global_Sales",
            "mean"
        ),
        Total_Global_Sales=(
            "Global_Sales",
            "sum"
        ),
        Average_NA_Sales=(
            "NA_Sales",
            "mean"
        ),
        Average_EU_Sales=(
            "EU_Sales",
            "mean"
        ),
        Average_JP_Sales=(
            "JP_Sales",
            "mean"
        ),
        Average_Other_Sales=(
            "Other_Sales",
            "mean"
        )
    )
    .reset_index()
)

print("\n================ CLUSTER SUMMARY ================\n")

display(cluster_summary)


# ============================================================
# 30. MOST COMMON GENRE IN EACH CLUSTER
# ============================================================

print("\n================ DOMINANT GENRE ================\n")

for cluster in sorted(
    df["Cluster"].unique()
):

    cluster_data = df[
        df["Cluster"] == cluster
    ]

    most_common_genre = (
        cluster_data["Genre"]
        .value_counts()
        .idxmax()
    )

    number_of_games = (
        cluster_data["Genre"]
        .value_counts()
        .max()
    )

    print(
        f"Cluster {cluster}: "
        f"{most_common_genre} "
        f"({number_of_games} games)"
    )


# ============================================================
# 31. TOP 10 GAMES FOR EVERY CLUSTER
# ============================================================

cluster_top_games = {}

for cluster in sorted(
    df["Cluster"].unique()
):

    cluster_top_games[cluster] = (
        df[
            df["Cluster"] == cluster
        ]
        .sort_values(
            "Global_Sales",
            ascending=False
        )
        .head(10)
        [
            [
                "Name",
                "Platform",
                "Genre",
                "Global_Sales"
            ]
        ]
    )

for cluster, games in cluster_top_games.items():

    print("\n")
    print("=" * 60)
    print(f"TOP GAMES - CLUSTER {cluster}")
    print("=" * 60)

    display(games)


# ============================================================
# 32. OPTIONAL: COMBINE NLP FEATURES WITH SALES FEATURES
# ============================================================

print(
    "\n================ NLP + SALES CLUSTERING ================\n"
)

# Standardise sales variables

sales_features = [
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales",
    "Global_Sales"
]

sales_data = (
    df[sales_features]
    .fillna(0)
)

scaler = StandardScaler()

X_sales = scaler.fit_transform(
    sales_data
)

print(
    "Sales feature matrix:",
    X_sales.shape
)


# ============================================================
# 33. REDUCE NLP FEATURES
# ============================================================

svd_50 = TruncatedSVD(
    n_components=50,
    random_state=42
)

X_text_reduced = (
    svd_50.fit_transform(
        X_tfidf
    )
)

print(
    "Reduced NLP feature matrix:",
    X_text_reduced.shape
)


# ============================================================
# 34. COMBINE NLP + SALES
# ============================================================

X_combined = np.hstack(
    [
        X_text_reduced,
        X_sales
    ]
)

print(
    "Combined feature matrix:",
    X_combined.shape
)


# ============================================================
# 35. FIND BEST K FOR COMBINED DATA
# ============================================================

combined_inertia = []
combined_silhouette = []

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(
        X_combined
    )

    combined_inertia.append(
        model.inertia_
    )

    combined_silhouette.append(
        silhouette_score(
            X_combined,
            labels
        )
    )


# ============================================================
# 36. COMBINED ELBOW PLOT
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(2, 11),
    combined_inertia,
    marker="o"
)

plt.title(
    "Elbow Method - NLP + Sales"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 37. COMBINED SILHOUETTE PLOT
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(2, 11),
    combined_silhouette,
    marker="o"
)

plt.title(
    "Silhouette Score - NLP + Sales"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 38. BEST K FOR COMBINED DATA
# ============================================================

best_combined_k = (
    range(2, 11)
)

best_combined_k = list(
    best_combined_k
)[
    np.argmax(
        combined_silhouette
    )
]

print(
    "Best K for combined NLP + Sales:",
    best_combined_k
)

print(
    "Best combined silhouette score:",
    round(
        max(combined_silhouette),
        4
    )
)


# ============================================================
# 39. FINAL COMBINED K-MEANS MODEL
# ============================================================

final_kmeans = KMeans(
    n_clusters=best_combined_k,
    random_state=42,
    n_init=10
)

df["Combined_Cluster"] = (
    final_kmeans.fit_predict(
        X_combined
    )
)

print(
    "Final NLP + Sales K-Means model completed."
)


# ============================================================
# 40. FINAL COMBINED CLUSTER SUMMARY
# ============================================================

combined_summary = (
    df.groupby("Combined_Cluster")
    .agg(
        Number_of_Games=("Name", "count"),
        Average_Global_Sales=(
            "Global_Sales",
            "mean"
        ),
        Total_Global_Sales=(
            "Global_Sales",
            "sum"
        ),
        Average_NA_Sales=(
            "NA_Sales",
            "mean"
        ),
        Average_EU_Sales=(
            "EU_Sales",
            "mean"
        ),
        Average_JP_Sales=(
            "JP_Sales",
            "mean"
        )
    )
    .reset_index()
)

display(combined_summary)


# ============================================================
# 41. FINAL CLUSTER COUNTS
# ============================================================

combined_counts = (
    df["Combined_Cluster"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(10, 6))

combined_counts.plot(
    kind="bar"
)

plt.title(
    "Number of Games in Each Combined Cluster"
)

plt.xlabel(
    "Combined Cluster"
)

plt.ylabel(
    "Number of Games"
)

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 42. FINAL CLUSTER SALES
# ============================================================

combined_sales = (
    df.groupby(
        "Combined_Cluster"
    )["Global_Sales"]
    .mean()
    .sort_values(
        ascending=False
    )
)

plt.figure(figsize=(10, 6))

combined_sales.plot(
    kind="bar"
)

plt.title(
    "Average Global Sales by Combined Cluster"
)

plt.xlabel(
    "Combined Cluster"
)

plt.ylabel(
    "Average Global Sales (Millions)"
)

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 43. SAVE FINAL DATASET
# ============================================================

output_file = (
    "Video_Games_Sales_NLP_KMeans_Results.csv"
)

df.to_csv(
    output_file,
    index=False
)

print(
    f"\nFinal dataset saved as: {output_file}"
)


# ============================================================
# 44. DISPLAY FINAL DATASET
# ============================================================

final_columns = [
    "Name",
    "Platform",
    "Year_of_Release",
    "Genre",
    "Publisher",
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales",
    "Global_Sales",
    "Cluster",
    "Combined_Cluster"
]

display(
    df[final_columns].head(50)
)


# ============================================================
# 45. FINAL REPORT
# ============================================================

print("\n")
print("=" * 70)
print("FINAL ANALYSIS REPORT")
print("=" * 70)

print(
    f"\nOriginal dataset rows: {len(pd.read_csv(file_path))}"
)

print(
    f"Cleaned dataset rows: {len(df)}"
)

print(
    f"\nNumber of NLP clusters: {best_k}"
)

print(
    f"NLP silhouette score: {best_score:.4f}"
)

print(
    f"\nNumber of NLP + Sales clusters: "
    f"{best_combined_k}"
)

print(
    f"NLP + Sales silhouette score: "
    f"{max(combined_silhouette):.4f}"
)

print(
    "\nFinal results have been saved to:"
)

print(
    output_file
)

print("\nAnalysis completed successfully.")
```
