```python
# ============================================================
# 1. Import libraries
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# NLP
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# K-Means and text vectorisation
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# Download NLP resources
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')


# ============================================================
# 2. Load the Olympic winners dataset
# ============================================================

df = pd.read_csv("olympic-winners.csv")

# Display first 5 rows
display(df.head())

# Dataset information
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())


# ============================================================
# 3. Basic data cleaning
# ============================================================

# Remove completely empty rows
df = df.dropna(how="all")

# Display column data types
print(df.dtypes)

# Fill missing values with an empty string
df = df.fillna("")

display(df.head())


# ============================================================
# 4. Find text columns for NLP
# ============================================================

text_columns = df.select_dtypes(include=["object"]).columns

print("Text columns:")
print(list(text_columns))


# ============================================================
# 5. Combine text columns
# ============================================================

# Combine all text columns into one NLP text field
df["combined_text"] = df[text_columns].astype(str).agg(" ".join, axis=1)

display(df[["combined_text"]].head())


# ============================================================
# 6. NLP preprocessing
# ============================================================

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    # Convert to lowercase
    text = text.lower()
    
    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    
    # Keep only letters and spaces
    text = re.sub(r"[^a-z\s]", " ", text)
    
    # Tokenise
    words = text.split()
    
    # Remove stopwords and very short words
    words = [
        word for word in words
        if word not in stop_words and len(word) > 2
    ]
    
    # Lemmatization
    words = [lemmatizer.lemmatize(word) for word in words]
    
    return " ".join(words)


df["clean_text"] = df["combined_text"].apply(clean_text)

display(df[["combined_text", "clean_text"]].head())


# ============================================================
# 7. Convert text into numerical features using TF-IDF
# ============================================================

vectorizer = TfidfVectorizer(
    max_features=1000,
    min_df=2,
    max_df=0.95
)

X = vectorizer.fit_transform(df["clean_text"])

print("TF-IDF matrix shape:", X.shape)


# ============================================================
# 8. Find a suitable number of K-Means clusters
# ============================================================

inertia = []
silhouette_scores = []

# Test different numbers of clusters
k_values = range(2, 11)

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X)
    
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X, labels))


# ============================================================
# 9. Elbow method
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(k_values, inertia, marker="o")

plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Choosing K")

plt.show()


# ============================================================
# 10. Silhouette score
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(k_values, silhouette_scores, marker="o")

plt.xlabel("Number of clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.show()

print("Silhouette scores:")

for k, score in zip(k_values, silhouette_scores):
    print(f"K = {k}: {score:.4f}")


# ============================================================
# 11. Apply K-Means
# ============================================================

# Change this value based on the elbow/silhouette results
optimal_k = 4

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

df["cluster"] = kmeans.fit_predict(X)

print(df["cluster"].value_counts().sort_index())


# ============================================================
# 12. Display Olympic winners by cluster
# ============================================================

for cluster in sorted(df["cluster"].unique()):
    
    print("\n" + "=" * 60)
    print(f"CLUSTER {cluster}")
    print("=" * 60)
    
    cluster_data = df[df["cluster"] == cluster]
    
    display(cluster_data.head(10))


# ============================================================
# 13. Find important words in each cluster
# ============================================================

terms = vectorizer.get_feature_names_out()

order_centroids = kmeans.cluster_centers_.argsort()[:, ::-1]

for cluster in range(optimal_k):
    
    print(f"\nCluster {cluster} important words:")
    
    top_words = [
        terms[ind]
        for ind in order_centroids[cluster, :15]
    ]
    
    print(", ".join(top_words))


# ============================================================
# 14. Visualise K-Means clusters using PCA
# ============================================================

# Convert sparse TF-IDF data into 2 dimensions
pca = PCA(n_components=2, random_state=42)

X_pca = pca.fit_transform(X.toarray())

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=df["cluster"],
    alpha=0.7
)

plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.title("K-Means Clustering of Olympic Winners")

plt.colorbar(scatter, label="Cluster")

plt.show()


# ============================================================
# 15. Save the clustered dataset
# ============================================================

df.to_csv(
    "olympic-winners-clustered.csv",
    index=False
)

print("Clustered dataset saved successfully.")


# ============================================================
# 16. Final summary
# ============================================================

print("Number of Olympic winners:", len(df))
print("Number of clusters:", optimal_k)

print("\nWinners in each cluster:")
print(df["cluster"].value_counts().sort_index())
```
